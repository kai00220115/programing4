# **数字当てゲーム**



## **概要**
　このプログラムはプレイヤーが難易度を選び，コンピュータが難易度に応じた範囲の数字からランダムに選んだ数字を当てるというシンプルなゲームです．また難易度が上がるとチャレンジできる回数も少なくなります．プレイヤーが数字を入力し，プログラムは正解の数字より大きいか小さいか，正解か教えてくれます．また入力した数字が正解の数字との誤差が５以下だった場合，知らせてくれます．制限された回数以内に正解を当てることができないと失敗します．そしてチャレンジ回数からスコアを出すことができます．入力した数字の履歴が画面上に表示され，もし同じ数字を誤って入力してしまっても警告が出されてチャレンジできる回数が減ってしまうこともありません．もし制限された回数以内に正解できなくても何度でも挑戦することができます．もう一度挑戦するかプログラムが聞いてくれます．このとき注意点が２つあります．まずもう一度挑戦するとき，ｙを入力すると挑戦できますが，半角で入力しないと終了してしまいます．次に難易度を選択するとき半角数字を入力しないと同じ難易度でもう一度プレイすることになってしまいます．

## **操作方法**

　PythonのコードをGoogle Colabの新しいコードセルに貼り付けて，コードセルの左側にある**実行ボタン**をクリックするか，shift + Enterを押して実行してください．コードが実行すると，難易度が選べます．1を入力して**easy**を選択しますと，数字の範囲が1から50で，15回チャレンジすることができます．2を入力し**normal**を選択しますと，1から100 の数字の範囲で10回チャレンジすることができます．3を入力し**hard**を選択しますと，1から200の数字の範囲で7回チャレンジすることができます．プロンプトが表示されたら，キーボードで1から各難易度ごとの範囲までの数字を入力し，Enterキーを押してください．入力した数字が正解と異なる場合，「もっと大きい数字です！」または「もっと小さい数字です！」と表示されます．また入力した数字が正解の数字との誤差が5以下だった場合「惜しい！」と表示されます．これらのヒントを参考に新しい数字を入力し再びEnterを押すと，正解に近づくことができます．正解の数字を当てると，「🎉 大正解！m 回目で当たったよ！(mはチャレンジした回数)」とメッセージが表示されチャレンジできる残りの回数からスコアを出すことができます．「もう一度遊ぶ？」と聞かれるのでもう一度遊ぶ場合，半角で**y**を入力するともう一度遊べます．このとき**n**を入力した場合，「またね！」とメッセージが表示され終了します．
## **制作者のコメント**
　難易度を選択できることが工夫したことです．また`list()`関数で入力した数字を表示することで数字をより当てやすくしたことです．そして大きい，小さいのヒントに加え誤差が５以下のとき特殊なヒントが生じることで当てやすくしました．私は今回の課題を通し，こういった自由課題が苦手なのだということが痛感しました．しかしこれからもっとこういった課題は増えるでしょうし，さらに社会に出ると参考にできるものはないといってもいいでしょう．そのためより技術を得るためにはもっと基礎的なことができないといけないということがわかりました．ネットで調べたものをもとにしていますが，先輩方が作ったものを一部参考にしているため，似ているところがあるかもしれませんがどうか許してください．

## **作成時間**

作成時間：**約15時間**



```python

import random

print("=== 数当てゲーム ===")

while True:
    #難易度
    print("\n難易度を選んでね！")
    print("1: easy (1〜50 / 15回)")
    print("2: normal (1〜100 / 10回)")
    print("3: hard(1～200 / 7)")

    mode = input("番号を選んで入力して!: ")
    if mode == "1":
        max_num = 50
        max_attempts = 15   #チャレンジできる回数
    elif mode == "2":
      max_num = 100
      max_attempts = 10
    elif mode == "3" :
        max_num = 200
        max_attempts = 7

    #ゲームの初期設定
    correct_number = random.randint(1, max_num)
    attempts = 0
    history = []  #入力した数字を覚えておくリスト
    is_correct = False

    print(f"\n1から{max_num}までの数字を当ててね！あなたには{max_attempts}回のチャンスがあります！")

    #メインのゲーム処理
    while attempts < max_attempts:
        try:
            guess = int(input(f"\n[{attempts + 1}回目] あなたの予測: "))
        except ValueError:
            print("無効な入力です")
            continue

        #リストを使って「同じ数字の二重入力」をチェック
        if guess in history:
            print("その数字はさっきも入力したよ！")
            continue

        #履歴に追加して、試行回数を1増やす
        history.append(guess)
        attempts += 1

        #判定処理
        if guess == correct_number:
            is_correct = True
            print(f"🎉 大正解！ {attempts} 回目で当たったよ！")

            #スコア計算
            score = (max_attempts - attempts + 1) * 10
            print(f"あなたのスコア: {score} 点！")
            break

        elif guess < correct_number:
            print("もっと大きい数字です！")
        else:
            print("もっと小さい数字です！")

        #差が5以内のとき
        if abs(guess - correct_number) <= 5:
            print("（惜しい！）")

        #これまでの入力履歴と残り回数を表示
        print(f"今までの入力: {history}")
        print(f"残りのチャンス: {max_attempts - attempts}回")

    #失敗したとき
    if not is_correct:
        print(f"\n正解は【 {correct_number} 】でした。")

    #リトライ機能
    retry = input("\nもう一度遊ぶ？ (y/n): ")
    if retry.lower() != 'y':
        print("またね！")
        break
```
